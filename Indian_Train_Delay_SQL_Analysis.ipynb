```python
import pandas as pd

df = pd.read_excel("/content/Indian_Train_Delay_Cleaned.xlsx")

df.head()
```

```python
import sqlite3

conn = sqlite3.connect("train_delay.db")
```

```python
df.to_sql(
    "train_delays",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT *
FROM train_delays
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    train_number,
    train_name,
    ROUND(AVG(average_delay_minutes), 2) AS avg_delay_minutes
FROM train_delays
GROUP BY train_number, train_name
ORDER BY avg_delay_minutes DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    train_number,
    train_name,
    ROUND(AVG(pct_right_time), 2) AS avg_right_time_pct
FROM train_delays
GROUP BY train_number, train_name
HAVING AVG(pct_right_time) > 90
ORDER BY avg_right_time_pct DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    station_code,
    station_name,
    ROUND(AVG(average_delay_minutes), 2) AS avg_delay_minutes,
    COUNT(*) AS record_count
FROM train_delays
GROUP BY station_code, station_name
HAVING COUNT(*) >= 10
ORDER BY avg_delay_minutes DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    station_code,
    station_name,
    ROUND(AVG(pct_right_time), 2) AS avg_right_time_pct,
    COUNT(*) AS record_count
FROM train_delays
GROUP BY station_code, station_name
HAVING COUNT(*) >= 10
ORDER BY avg_right_time_pct DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    train_number,
    train_name,
    average_delay_minutes,
    CASE
        WHEN average_delay_minutes IS NULL THEN 'No Data'
        WHEN average_delay_minutes = 0 THEN 'On Time'
        WHEN average_delay_minutes <= 15 THEN 'Slight Delay'
        WHEN average_delay_minutes <= 60 THEN 'Moderate Delay'
        ELSE 'High Delay'
    END AS delay_category
FROM train_delays
LIMIT 20;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    CASE
        WHEN average_delay_minutes IS NULL THEN 'No Data'
        WHEN average_delay_minutes = 0 THEN 'On Time'
        WHEN average_delay_minutes <= 15 THEN 'Slight Delay'
        WHEN average_delay_minutes <= 60 THEN 'Moderate Delay'
        ELSE 'High Delay'
    END AS delay_category,
    COUNT(*) AS record_count
FROM train_delays
GROUP BY delay_category
ORDER BY record_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    train_number,
    train_name,
    ROUND(AVG(average_delay_minutes), 2) AS avg_delay_minutes,
    RANK() OVER (
        ORDER BY AVG(average_delay_minutes) DESC
    ) AS delay_rank
FROM train_delays
GROUP BY train_number, train_name;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT *
FROM (
    SELECT
        train_number,
        train_name,
        ROUND(AVG(average_delay_minutes), 2) AS avg_delay_minutes,
        RANK() OVER (
            ORDER BY AVG(average_delay_minutes) DESC
        ) AS delay_rank
    FROM train_delays
    GROUP BY train_number, train_name
)
WHERE delay_rank <= 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    COUNT(*) AS total_records,
    COUNT(DISTINCT train_number) AS unique_trains,
    COUNT(DISTINCT station_code) AS unique_stations,
    ROUND(AVG(average_delay_minutes), 2) AS avg_delay_minutes,
    ROUND(AVG(pct_right_time), 2) AS avg_right_time_pct,
    ROUND(AVG(pct_slight_delay), 2) AS avg_slight_delay_pct,
    ROUND(AVG(pct_significant_delay), 2) AS avg_significant_delay_pct,
    ROUND(AVG(pct_cancelled_unknown), 2) AS avg_cancelled_unknown_pct
FROM train_delays;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    ROUND(MAX(average_delay_minutes), 2) AS maximum_delay,
    ROUND(MIN(average_delay_minutes), 2) AS minimum_delay
FROM train_delays;
"""

pd.read_sql_query(query, conn)
```
